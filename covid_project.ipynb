```python

cases_cont=df.groupby("continent")["Cases"].sum()
plt.pie(cases_cont,labels=cases_cont.index,autopct="%1.1f%%",colors=["red","blue","green","yellow","pink","grey","Purple"])
plt.title("cases per continent")
plt.figure(figsize=(10,7))
plt.show()
```

```python
x='Cases'
y='country'
group=df.groupby(y)[x].sum().sort_values(ascending=False).head(10).reset_index()
plt.title("top 10 countries with most cases")
sns.barplot(x=x,y=y,data=group)
```

```python
plt.figure(figsize=(10,9))
p="population"
c="country"
pop=df.groupby(p)[c].sum().sort_values(ascending=False).head(5).reset_index()
sns.barplot(x=c, y=p, data=pop)
plt.title("5 countries with heighest population")
```

```python
x='Deaths'
y='country'
group=df.groupby(y)[x].sum().sort_values(ascending=False).head(10).reset_index()
sns.barplot(x=x,y=y,data=group)
```

```python

rec_country = (df.groupby("country")["Recovered"].sum().sort_values(ascending=False).head(5).reset_index())
plt.barh(rec_country["country"], rec_country["Recovered"],color=["blue","red","green","gray","yellow"])
plt.xlabel("sum of recovered cases")
plt.ylabel("countries")
plt.title("countries with heighest recovered patients")
plt.show()
```

```python

plt.figure(figsize=(8,5))
continent_rec = df.groupby("continent")["Recovered"].sum().reset_index()
plt.plot(continent_rec["continent"], continent_rec["Recovered"], marker="o", linewidth=2)
plt.xlabel("Continent")
plt.ylabel("Recovered Cases")
plt.title("Total Recovered Cases by Continent")
plt.grid(True)
plt.show()
```

```python
# Make sure no division by zero errors
df["Cases"] = df["Cases"].replace(0, 1)

# Death Rate
df["Death Rate"] = df["Deaths"] / df["Cases"]

# Recovery Rate
df["Recovery Rate"] = df["Recovered"] / df["Cases"]

# Convert to percentage (optional but better for visuals)
df["Death Rate (%)"] = df["Death Rate"] * 100
df["Recovery Rate (%)"] = df["Recovery Rate"] * 100

# Check results
print(df[["country", "Cases", "Deaths", "Recovered", "Death Rate (%)", "Recovery Rate (%)"]].head())
```
